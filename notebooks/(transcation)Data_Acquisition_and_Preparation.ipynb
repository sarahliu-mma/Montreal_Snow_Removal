```python
dt['DateChargement'] = pd.to_datetime(dt['DateChargement'],errors='coerce')
dt['DateDechargement'] = pd.to_datetime(dt['DateDechargement'],errors= 'coerce')
```

```python
print(dt['DateChargement'].min())
print(dt['DateChargement'].max())
```

```python
#Abnormal observation
date_is_abnormal = (dt['DateChargement'] < '2025-11-01') | (dt['DateChargement'] > '2026-04-30')
volume_is_abnormal = dt['volume'] == 0
```

```python
print(date_is_abnormal.sum())
print(volume_is_abnormal.sum())
```

```python
dt = dt[~volume_is_abnormal]
print(dt.shape)
```

```python
dt = dt[~date_is_abnormal]
print(dt.shape)
```

```python
#filter out the transcations that werent operated by the government
n_official_operation0 = dt['NomSecteur'].isna()
n_official_operation1 = dt['NumeroContrat'].str.contains('EXT',na= False)
n_official_operation2 = dt['NumeroContrat'].isna()
```

```python
print(n_official_operation0.sum())
print(n_official_operation1.sum())
print(n_official_operation2.sum())
```

```python
dt = dt[~n_official_operation0]
dt = dt[~n_official_operation1]
dt = dt[~n_official_operation2]
```

```python
dt.shape
```

```python
ct = pd.crosstab(dt["TypeTransaction"],n_official_operation0)
print(ct)
```

```python
dt = dt[~no_date]
```

```python
too_long = (dt['DateDechargement'] - dt['DateChargement']) > pd.Timedelta(hours=6)
dt.loc[too_long, 'DateChargement'] = dt.loc[too_long, 'DateDechargement']
```

```python
url = ("https://climate.weather.gc.ca/climate_data/bulk_data_e.html"
       "?format=csv&stationID=51157&Year={year}&Month=1&Day=1"
       "&timeframe=2&submit=Download+Data")

w2025 = pd.read_csv(url.format(year=2025))
w2026 = pd.read_csv(url.format(year=2026))
weather = pd.concat([w2025, w2026], ignore_index=True)
```

```python
weather['Date/Time'] = pd.to_datetime(weather['Date/Time'])
```

```python
print(weather['Date/Time'].min())
print(weather['Date/Time'].max())
```

```python
date_filter = (weather['Date/Time'] >= '2025-10-01') & (weather['Date/Time'] <= '2026-04-30')
```

```python
weather = weather[date_filter]
```

```python
weather = weather[['Date/Time','Max Temp (°C)','Min Temp (°C)','Mean Temp (°C)','Total Rain (mm)','Total Snow (cm)','Total Precip (mm)','Snow on Grnd (cm)']]
```

```python
weather = weather.copy()
weather = weather.sort_values('Date/Time')

weather['Snow on Grnd (cm)'] = weather['Snow on Grnd (cm)'].ffill().fillna(0)
```

```python
weather.isna().sum()
```

```python
for col in ['Max Temp (°C)', 'Min Temp (°C)', 'Mean Temp (°C)']:
    weather[col] = weather[col].interpolate()
```

```python
for col in ['Total Rain (mm)', 'Total Snow (cm)', 'Total Precip (mm)']:
    weather[col] = weather[col].fillna(0)
```

```python
weather.isna().sum()
```

```python
weather = weather.rename(columns={
    'Date/Time': 'date',
    'Max Temp (°C)': 'tmax',
    'Min Temp (°C)': 'tmin',
    'Mean Temp (°C)': 'tmean',
    'Total Rain (mm)': 'rain',
    'Total Snow (cm)': 'snow',
    'Total Precip (mm)': 'precip',
    'Snow on Grnd (cm)': 'snow_grnd',
})
```
